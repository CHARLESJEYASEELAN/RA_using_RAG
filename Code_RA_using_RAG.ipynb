🛠️ Step 1: Improve Document Chunking
By default, LangChain splits PDFs into chunks, but we can optimize the chunking process.
Try Text Splitters for better context retention.

🔹 Modify your existing script with RecursiveTextSplitter:

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader 

#load PDF
pdf = PyPDFLoader("1.pdf")
docs = pdf.load() 

#Optimized Chunking than the default chunking level!! 
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)
docs = text_splitter.split_documents(docs)
```

🛠️ Step 2: Rank Most Relevant Information Using Embeddings

Sometimes, retrieved chunks are not always perfect.
We’ll use similarity scoring to rank retrieved documents.

```python
from langchain_community.llms import Ollama 
from langchain_community.embeddings import OllamaEmbeddings 
from langchain_community.vectorstores import FAISS 
from langchain.chains import RetrievalQA
```

```python
llm = Ollama(model='mistral')
```

```python
embedding = OllamaEmbeddings(model='mistral')
vectordb = FAISS.from_documents(docs,embedding)
```

```python
retriever = vectordb.as_retriever(search_kwargs={"k":3}) # top 3 relevant chunks
qa_chain = RetrievalQA.from_chain_type(llm,retriever=retriever) 
```

```python
while True:
    query = input("\n🗣️ Ask me anything about your research paper: ")
    if query.lower() in ["exit", "quit"]:
        print("👋 Exiting...")
        break
    response = qa_chain.invoke({"query": query})
    print("\n🤖 AI:", response["result"])
```
